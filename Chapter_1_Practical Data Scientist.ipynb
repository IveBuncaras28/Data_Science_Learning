# What a Notebook Actually Is
A Jupyter Notebook is a document format, saved with the .ipynb extension, that interleaves three distinct kinds of content in a single, linear file: executable code, the output that code produces, and narrative text written in Markdown (Kluyver et al., 2016). Unlike a text editor showing a .py file, or a spreadsheet showing cells of values and formulas, a notebook is structured as a sequence of discrete units called cells. Each cell can be run independently, and its result – a printed value, a table, a chart, or an error – appears directly beneath it, permanently attached to the code that produced it.

This architecture traces back to IPython, the interactive Python shells developed by Fernando Pérez in 2001 and formalized in Pérez and Granger’s (2007) foundational paper on interactive scientific computing. IPython’s core insight was that scientific and exploratory programming does not proceed like traditional software engineering – in a straight line from requirements to finished product – but iteratively, through cycles of running a small piece of code, inspecting the result, and adjusting course. The notebook interface, introduced later as part of the IPython project and eventually spun off into the language-agnostic Project Jupyter in 2014, extended that same interactive philosophy into a persistent, shareable document (Kluyver et al., 2016). The name itself – Jupyter – is a reference to the three core languages the project originally supported: Julia, Python, and R, though it now supports over 100 languages through its kernel architecture (Project Jupyter, n.d.).


# Why This Differs from a Script
A traditional Python script is written for the computer to execute from top to bottom, all at once, with no visible intermediate state unless the programmer explicitly writes print( ) statements. If something goes wrong on line 340, the developer typically cannot see what the data looked like on line 200 without re-running the entire script or attaching a debugger. This model suits software that is finished – code whose logic has already been worked out and is now being deployed. It does not suit the earlier, messier phase of data science, where the practitioner does not yet know what the data contains, which approach will work, or what the “correct” logic even is.

A notebook inverts this. Because each cell can be executed and inspected on its own, a data scientist can load a dataset in one cell, look at its first few rows in the next, test a cleaning step in a third, and only afterwards decide how those steps should be assembled into a final pipeline. Rule et al. (2018), in their empirical study of how practitioners actually use computational notebooks, describe this as a distinct workflow combining exploration – rapid, provisional, often abandoned experimentation – with explanation – the retrospective narrative built to communicate what was learned. Scripts are built to support both within the same artifact.


# Why This Differs from Excel
Spreadsheet software such as Excel solves a related but different problem. It excels at direct manipulation of small, structured, rectangular data through formulas and visible cell references, and it has long been the default tool for exploratory data work outside of specialized programming environments. However, three limitations become apparent once a dataset or an analysis grows beyond a trivial size.

First, an Excel workbook does not, by default, retain an explicit, inspectable record of the sequence of transformations applied to the data — reproducing an analysis (or auditing it) often means manually tracing formulas across a grid of cells rather than reading a linear log of steps. Second, spreadsheet formulas are difficult to test, version, and reuse: a single copy-paste error can silently propagate incorrect results across thousands of cells with no visible trace (a well-documented failure mode discussed extensively in the spreadsheet-error literature, e.g., Panko, 1998). Third, Excel has a hard ceiling on both data volume and the sophistication of the analysis it can perform — statistical modeling, machine learning, or natural language processing are largely outside its native capabilities, whereas a notebook has direct access to the full Python (or R, or Julia) ecosystem of libraries.

A notebook, by contrast, keeps the entire analytical logic as readable, re-runnable code, checked into cells in an explicit order, with the narrative explaining why a step was taken written directly alongside the step itself. This combination — code, output, and explanation, in one place — is precisely what Rule et al. (2018) and later Pimentel et al. (2019) point to as the notebook's central affordance: it functions simultaneously as an analysis tool, a lab notebook, and a communication medium, collapsing three separate documents into one. 


# Why the Combination Matters for Data Science Specifically

It is worth being precise about why this matters for data science in particular, rather than treating it as a generic productivity claim. Data science is fundamentally an evidentiary discipline: a claim such as "sales are increasing" or "this segment of users is at higher churn risk" is only as credible as the chain of transformations and decisions that produced it. A notebook makes that chain visible and auditable in a way that neither a black-box script nor an opaque spreadsheet grid does — a reviewer, a collaborator, or the author's own future self can read the notebook top to bottom and see not just the final number, but the reasoning that led there. This transparency is also precisely where notebooks can fail if used carelessly (cells run out of order, hidden state, unreferenced files), a set of pitfalls documented at scale by Pimentel et al. (2019) in their empirical study of over one million public Jupyter notebooks — and a set of pitfalls this book addresses directly in Chapter 3.

# A Shared Foundation, Four Different Front Ends
Before comparing these tools, it is important to understand what they actually are: none of them is a competing "notebook technology." All four are different interfaces built on top of the same underlying open standard — the .ipynb file format and the Jupyter messaging protocol that allows a front-end interface to communicate with a back-end computational engine, or kernel (Kluyver et al., 2016). This distinction matters because it means a notebook created in one environment can, in principle, be opened and run in any of the others; what differs is the surrounding experience — the file browser, the extensions available, where the computation actually happens, and how easy it is to share the result.

# Classic Notebook and JupyterLab
The original web-based notebook interface, released as part of the IPython project and later renamed the "classic Jupyter Notebook," presented a minimal, single-document view: one notebook, one browser tab, little else (Kluyver et al., 2016). JupyterLab, released as the project's next-generation interface starting in 2018, retained the same notebook document model but embedded it inside a full integrated development environment — a file browser, multiple tabs, a text editor, a terminal, and an extension system, all in one window (Project Jupyter, n.d.-a). As of the Jupyter project's own documentation, JupyterLab is now positioned as the primary recommended interface, with the classic Notebook interface maintained largely for users who prefer its simplicity or have workflows built around it (Project Jupyter, n.d.-b). For a learner starting today, this practical convergence is why this book, like most current instructional material, defaults to JupyterLab: it offers strictly more capability with no meaningful cost in the basic case of running a single notebook.

# Google Colab
Google Colaboratory ("Colab") is a hosted, browser-based notebook environment built by Google that runs entirely on Google's infrastructure rather than the user's own machine (Bisong, 2019). Functionally, it presents the same cell-based .ipynb interface as JupyterLab, but with three defining differences. First, it requires no local installation of Python or any libraries — a user with only a browser and a Google account can begin writing and running code within seconds, which removes what Guo (2018) identifies, in a study of novice programmers' setup struggles, as one of the largest early barriers to learning to program: environment configuration failures unrelated to the actual skill being taught. Second, Colab provides free, time-limited access to hardware accelerators — GPUs and TPUs — which is particularly relevant for the deep learning material in Chapter 14 of this book, where local hardware may otherwise be a limiting factor (Bisong, 2019; Carneiro et al., 2018). Third, because notebooks are stored on Google Drive by default, Colab is built around real-time collaborative editing in a manner comparable to Google Docs, which is not a native feature of a locally run JupyterLab session.

These conveniences come with corresponding tradeoffs. Colab sessions run on temporary virtual machines that are recycled after a period of inactivity, meaning locally installed packages, uploaded files, and any variables held in memory do not persist between sessions unless explicitly saved to Google Drive or reloaded at the start of each run (Bisong, 2019; Google, n.d.). Data uploaded to a Colab notebook is transmitted to and processed on Google's servers, which raises legitimate data governance and confidentiality considerations for anyone working with sensitive, proprietary, or human-subjects data — a concern directly relevant to a study, such as this book's author's own research involving adolescent participants, where data handling is typically governed by an institutional ethics protocol (Bisong, 2019). Finally, extended or GPU-intensive use is subject to usage limits under the free tier, with guaranteed resources available only through paid subscription tiers (Google, n.d.).

# VS Code Notebooks
Visual Studio Code, Microsoft's general-purpose source code editor, added native support for opening and running .ipynb files directly within its own interface via the Jupyter extension (Microsoft, n.d.). This allows a developer to work on notebooks using the same editor, keyboard shortcuts, extensions, and integrated debugging tools they may already use for writing production Python code, scripts, or larger software projects — a workflow advantage specifically for users who move frequently between exploratory notebook work and conventional software development (Microsoft, n.d.). Under the hood, VS Code still connects to a standard Jupyter kernel — either one installed locally via Miniconda/Anaconda, or a remote kernel — meaning it is best understood not as a fourth, independent notebook technology but as an alternative, IDE-embedded front end to the same underlying execution model described above (Kluyver et al., 2016).

# The Dependency Problem Miniconda Solves
Before installing anything, it is worth understanding the problem an environment manager like Miniconda actually solves, because the reasoning explains an otherwise mysterious step (creating a separate "environment" before installing packages). Python data science work depends on a stack of third-party libraries — pandas, NumPy, scikit-learn, and others — each of which is itself under active development and released in successive versions. A project built against pandas version 1.5 may break, subtly or completely, if later run against pandas version 2.2, because function signatures, default behaviors, and even entire APIs can change between major versions (McKinney, 2022). If every project on a machine shares one global Python installation, upgrading a library for one project risks silently breaking another — a failure mode informally known in the software engineering community as "dependency hell," and one of the most commonly cited reproducibility failures in computational research (Sandve et al., 2013).

Conda, the package and environment manager originally developed for the Anaconda distribution and made available in a minimal form as Miniconda, addresses this by allowing a user to create isolated, self-contained environments, each with its own Python interpreter and its own independently versioned set of packages (Anaconda, Inc., n.d.). Critically, Conda manages not only Python packages but also underlying non-Python dependencies (compiled libraries, system-level binaries) that many scientific packages rely on — a broader scope than Python's built-in pip and venv tools, which is part of why Conda became the de facto standard in the scientific Python community (Millman & Pérez, 2014). Miniconda specifically installs only Conda itself and Python, rather than the roughly 250 additional packages bundled into the full Anaconda distribution, making it the leaner and more widely recommended starting point for a new environment (Anaconda, Inc., n.d.).


# Why Reproducibility Is Not a Side Concern
This is not merely a technical convenience; it goes to the credibility of the resulting analysis. Sandve et al. (2013), in an influential set of rules for reproducible computational research, list precisely tracking the software environment used to produce a result as one of the foundational requirements for any analysis intended to be trusted or repeated — by others, or by the original author months later. An environment file (environment.yml) generated from a Conda environment functions as a machine-readable record of exactly which library versions produced a given result, in the same spirit as a materials-and-methods section in a scientific paper. This book returns to this concept formally in Chapter 3, but the practice begins here, at installation.


# Installing Locally with Miniconda

The installation itself follows three steps, corresponding to the exercise at the end of this section. First, the Miniconda installer is downloaded for the user's specific operating system from the official distribution site and run with default settings (Anaconda, Inc., n.d.). Second, a dedicated environment is created for this book's work, specifying the Python version explicitly so that it remains fixed regardless of what is installed system-wide:

```bash
conda create -n datasci python=3.11
conda activate datasci
```

Third, the core libraries used throughout this book — JupyterLab itself, along with pandas, NumPy, Matplotlib, Seaborn, and scikit-learn — are installed into that isolated environment:

```bash
pip install jupyterlab pandas numpy matplotlib seaborn scikit-learn
```

Note that pip, Python's native package installer, is used here inside an already-isolated Conda environment; this is standard, widely documented practice, since the isolation Conda provides at the environment level means pip-installed packages remain contained to that environment rather than affecting the system Python (Anaconda, Inc., n.d.). Once installed, JupyterLab is launched with a single command:

```bash
jupyter lab
```

This starts a local web server on the user's own machine and opens the JupyterLab interface in the default browser — a point worth making explicit, because it clarifies that Jupyter is, technically, a local website rather than a conventional desktop application, which is why it opens in a browser tab at all (Kluyver et al., 2016).

# Colab as the Zero-Install Alternative
Google Colab, introduced in the previous section, sidesteps this entire installation process by pre-provisioning a remote environment with Python and many common data science libraries already installed on Google's infrastructure (Bisong, 2019). This is the appropriate choice when the goal is to begin writing and running code immediately, without administrative rights on a machine, without a compatible operating system configuration, or without the time investment of a local setup — circumstances Guo (2018) identifies as disproportionately affecting learners without prior systems-administration experience. The tradeoff, as discussed in Section 1.2, is that a Colab environment is temporary and cloud-hosted: it does not produce a durable, versionable local environment file in the way a Conda environment does, and it is not the appropriate choice once a project involves data that must remain strictly on local or institutional infrastructure for privacy or ethical reasons (Bisong, 2019).


# Choosing Between Them for This Book
In practice, the two paths are not mutually exclusive, and the choice can reasonably change from project to project. A learner without administrative access to a shared or institutional computer, or one who wants to test the very first exercises in this chapter before committing to any installation, should start in Colab. A learner planning to work with local files, larger datasets, or data subject to a confidentiality agreement — a common situation in academic research contexts — should install locally via Miniconda from the outset, so that the environment-management habits in this section, and the reproducibility practices formalized in Chapter 3, are in place from the first notebook onward.

# What a Kernel Is
A Jupyter kernel is the separate computational process that actually executes the code contained in a notebook's cells; the notebook interface itself — whether JupyterLab, Colab, or VS Code — is only the front end that sends code to the kernel and displays whatever the kernel sends back (Kluyver et al., 2016). This separation is deliberate and is formalized in the Jupyter messaging protocol, which defines a language-agnostic communication format between front end and kernel over a set of ZeroMQ sockets (Pérez & Granger, 2007; Kluyver et al., 2016). Because the protocol is language-agnostic, a kernel does not have to run Python at all — the same notebook interface can drive kernels for R, Julia, or over a hundred other languages, provided a kernel implementing the protocol exists for that language (Project Jupyter, n.d.). For this book, the relevant kernel is IPython, the reference implementation that executes Python code and is installed automatically alongside JupyterLab.

# Why the Kernel Remembers Things
The behavior that most surprises newcomers follows directly from this architecture: a kernel is a single, continuously running process that persists for the lifetime of the notebook session, and it retains every variable, import, and function definition from every cell run so far, regardless of the order in which those cells appear on the page (Kluyver et al., 2016). Running cell 5 and then cell 2 executes them in that order as far as the kernel is concerned — the kernel has no awareness of a cell's position in the document, only the order in which it received code. This is precisely the property that Rule et al. (2018) identify as enabling the exploratory, non-linear workflow described in Section 1.1: a data scientist can jump back, redefine a variable, and re-run a downstream cell without restarting from scratch.

The same property, however, is the leading cause of a well-documented class of reproducibility failures. Pimentel et al. (2019), in their large-scale empirical study of over one million publicly shared Jupyter notebooks, found that a majority could not be executed top-to-bottom to reproduce their own displayed output, and identified out-of-order execution — a notebook whose visible cell numbers no longer reflect the order they were actually run in — as one of the most common underlying causes. A notebook can display a correct-looking result from a variable that was, in fact, set by a cell that has since been edited or deleted, leaving no visible trace of the discrepancy on the page.

# Why Restarting Is the Correct Fix
Restarting a kernel discards this entire accumulated state and starts a fresh process with no variables, no imports, and no memory of anything previously run (Kluyver et al., 2016). This is why "restart the kernel" is the standard first troubleshooting step recommended across Jupyter's own documentation and widely echoed in data science pedagogy: a bug that disappears after a restart-and-rerun was never a bug in the visible code at all, but a symptom of hidden state — a variable set earlier, in a cell that may no longer even exist in its original form.

The more rigorous version of this practice, "Restart & Run All," clears the kernel and then executes every cell from the first to the last in strict document order. Passing this check is functionally equivalent to what Sandve et al. (2013) describe as a minimal reproducibility test: verifying that the analysis, as written, actually produces the results it claims to produce, from a clean starting state, with no dependence on steps that happened to be run earlier and never repeated. This book treats "Restart & Run All completes without error" as the baseline acceptance criterion for every exercise notebook, and returns to this practice formally as part of the reproducibility habits built in Chapter 3.

# Two Cell Types, Two Purposes
Every unit of content in a Jupyter notebook lives inside a cell, and each cell has a type that determines how the kernel treats it when executed. A “code cell” contains source code — Python, in this book's case — that is sent to the kernel for execution when the cell is run, with any output (a printed value, a table, an error, a plot) displayed immediately beneath it (Kluyver et al., 2016). A “Markdown cell”, by contrast, is never sent to the kernel at all; instead, when run, it is rendered directly in the browser as formatted text, using Markdown, a lightweight plain-text formatting syntax originally created by John Gruber in 2004 specifically so that plain text could be both human-readable in its raw form and convertible to structured HTML (Gruber, 2004). Within a Markdown cell, a line beginning with `##` renders as a heading, text wrapped in “asterisks” renders as bold, and a line beginning with `-` renders as a bulleted list item — the same lightweight conventions used across GitHub, documentation sites, and this book's own source files.

# The Idea Behind Mixing the Two
The decision to let a single document interleave executable code with formatted prose is not a Jupyter-specific novelty; it is a direct descendant of “literate programming”, a paradigm introduced by Donald Knuth (1984) on the premise that a program should be written primarily as an explanation, readable by a human, with the machine-executable code embedded within that explanation rather than the reverse. Knuth's original framing — that the programmer's central task is to explain to other humans what they want the computer to do — maps almost exactly onto what a well-constructed notebook does in a data science context: a Markdown cell states the question being asked or the finding just observed, and the code cell beneath it carries out exactly that step. Rule et al. (2018) identify this interleaving as central to how practitioners actually use notebooks in practice, describing the resulting document as serving simultaneously as an executable analysis and a readable narrative — precisely the "explanation" function that a bare script or a spreadsheet, discussed in Section 1.1, does not natively support.

# Running Cells with Shift + Enter
Mechanically, a cell of either type is executed by selecting it and pressing “Shift + Enter”, which runs the current cell and automatically advances the selection to the next cell (creating a new one if none exists) — the single most-used command in notebook work, and the one this book assumes as the default rhythm for working through every exercise (Project Jupyter, n.d.). Two close variants are worth knowing early: “Ctrl/Cmd + Enter” runs the current cell but keeps it selected, useful when re-running the same cell repeatedly while adjusting a parameter; and “Alt/Option + Enter” runs the current cell and inserts a brand-new cell immediately after it, useful when actively writing a new sequence of steps.

The habit this book asks readers to build from Chapter 1 onward is straightforward: before writing a new code cell, write a short Markdown cell above it stating, in one sentence, what question that code answers or what step it performs. This is not a stylistic flourish. Pimentel et al. (2019) found that the presence of explanatory Markdown correlates with notebooks that are substantially easier for a second reader to understand and successfully re-run — a direct, measurable benefit of the same literate-programming discipline Knuth (1984) originally proposed for conventional software.

# What Pandas Is and Why It Exists
Pandas is an open-source Python library that provides data structures and operations for working with structured, tabular data — data organized into rows and labeled columns, the same shape as a spreadsheet or a database table (McKinney, 2010). It was created by Wes McKinney beginning in 2008 while working in financial analytics, specifically to close a gap he identified in the Python ecosystem at the time: Python had strong general-purpose numerical computing through NumPy, but lacked a structure suited to the messier, heterogeneously typed, labeled data that dominates real-world analysis — data with named columns, mixed types, and missing values, as opposed to NumPy's uniform numerical arrays (McKinney, 2010, 2022). The core object pandas introduced to fill this gap is the “DataFrame”, a two-dimensional, labeled data structure explicitly modeled on the data frame concept already established in the R programming language, adapted to integrate with Python's broader scientific computing stack (McKinney, 2010).

The significance of this design choice is not merely technical convenience. Wickham (2014), in an influential paper on data structuring conventions, argues that a large proportion of the effort in any real analysis is spent not on statistics or modeling but on reshaping and cleaning data into a consistent, well-structured form — a phase he terms "tidying." A DataFrame, with its explicit row and column labels, its per-column data types, and its built-in handling of missing values, gives that tidying work a concrete, consistent structure to target, which is a substantial part of why pandas became, and remains, the de facto standard for tabular data manipulation in the Python data science ecosystem (McKinney, 2022; Reback et al., 2020).

# Reading a CSV into a DataFrame
The specific function used to load a CSV file,  pandas.read_csv(), parses a delimited text file and returns it as a DataFrame, automatically inferring a sensible data type for each column (numeric, text, date-like) from the values it contains (McKinney, 2022). In this book, importing the library and loading the first NorthWind Retail file looks like this:

```python
import pandas as pd

orders = pd.read_csv("orders.csv")
```

The `pd` alias is not a formal requirement of the library, but it is a near-universal convention throughout the pandas documentation, published tutorials, and the wider open-source ecosystem (McKinney, 2022; Reback et al., 2020) — a convention worth adopting from the first line of code in this book simply because it makes every future pandas example, in this text or elsewhere, immediately legible.

# Inspecting the Result with .head()
Once loaded, a DataFrame can, in principle, contain anywhere from a handful of rows to many millions, which makes displaying the entire object neither useful nor practical as a first step. The .head() method addresses this directly: called on a DataFrame, it returns only the first “n” rows (five, by default, though any number can be passed explicitly, e.g.,    orders.head(10), rendered as a formatted table showing every column, its header, and a representative sample of its values (McKinney, 2022). This single call is, in practice, the near-universal first action taken on any newly loaded dataset, because it answers the most immediate structural questions — what columns exist, what their values look like, whether the data loaded as expected at all — before any cleaning, analysis, or visualization is attempted. McKinney (2022) presents this same load-then-inspect sequence as the canonical entry point to virtually every data analysis task in the pandas ecosystem, and this book adopts it as the standard opening move for every new dataset introduced from this chapter forward, beginning with NorthWind Retail's `orders.csv` in the hands-on exercise that follows.
